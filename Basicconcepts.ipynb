# Probability distribution in quant finance
Probability density function:
$$ P(a \le X \le b)= ∫_a^bP(x)dx $$


Skewness $S$:
$$
S=
\langle (x-m)^3/\sigma^3 \rangle
$$
About symmetry with respect to the mean value.
Kurtosis $K$:
$$ K = \langle (x-m)^4/\sigma^4 \rangle $$
Characterizes how peaky the peak is. Measurement of how much we deviate from normal distribution
