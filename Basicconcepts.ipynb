### 1.1 Lognormal distribution

What is it= Continuous probability distribution of a random variable whose logarithm is normally distributed. If X is lognormal, means:
$$ Y= lnX \sim N(\mu, \sigma^2) $$


Density function:
$$f(x; \mu, \sigma) = \frac{1}{x \sigma \sqrt{2\pi}} \exp\left(-\frac{(\ln x - \mu)^2}{2\sigma^2}\right)$$

Key facts:
*   takes only positive real values

Sigma meaning: meaning of standard deviation


*   Measures volatility/uncertainty
*   Normal dist: changes the width of the curve. Lognormal dist: changes its form
* $\sigma → 0$: closer to your typical normal curve. Low volatility, very stable
* $\sigma >> 1$: long tail to the right. High risk scenario.  






```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm


x = np.linspace(0.01, 5.0, 500) ## index set positive

# 3 sigma values to compare
mu = 0.0
sigmas = [0.3, 0.6, 1.0]

plt.figure(figsize=(6, 4))

for sigma in sigmas:
    # s: standard deviation of log
    # scale = exp(mu)
    scale = np.exp(mu)
    pdf = lognorm.pdf(x, s=sigma, scale=scale)
    plt.plot(x, pdf, label=f'$\\sigma$ = {sigma}', linewidth=2.5)

plt.title('Comparación $\\sigma$ ($\\mu = 0$)', fontsize=14)
plt.xlabel('Valores de la variable ($X$)', fontsize=12)
plt.ylabel('Densidad de Probabilidad $f(x)$', fontsize=12)
plt.axvline(np.exp(mu), color='gray', linestyle='--', alpha=0.7, label=f'Mediana ($e^\\mu$ = {np.exp(mu)})')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)
plt.ylim(bottom=0)

plt.tight_layout()
plt.show()
```

### 2.3 Markov process

Sum up: stochastic process describing a sequence of possible events in which the probability of each event depends only on the state attained in the previous event. All past info is useless.
$f(x_i): $ prob dist function for different values $x_i$ at different times $t_i$. Also used $P(x_i)$.


**Starting point:** in the most basic stochastic process the present has no dependence on the past. The conditional probability density function $f$ is:
$$
f(x_1,t_1; x_2,t_2;...)= \prod_i f(x_i,t_i)
$$

So the probs are independent and the probability of the values $x_1, x_2, x_3$ ocurring simultaneously is the product of each probability of each value.

**Markov process:** next level unlocked -> the future depends on the present.

$$f(x_1; x_2; ... \mid x_{k+1}; x_{k+2}; ...) = f(x_1; x_2; ...\mid X(t_{n-1}) = x_{n-1})$$

Meaning: evolution of the system is determined by the initial condition.

**Chapman-Kolmogorov equation:**
The prob of going (transitioning) from an initial state ($x_1, t_1$) to ($x_3, t_3$) can be obtained integrating on all the possible intermediate states ($x_2, t_2$).

$$P(x_3, t_3 \mid x_1, t_1) = \int_{-\infty}^{\infty} P(x_3, t_3 \mid x_2, t_2) P(x_2, t_2 \mid x_1, t_1) \, dx_2$$

Another notation (more intuitive imo):
$$p(x, t + \Delta t \mid x_0, t_0) = \int_{-\infty}^{\infty} p(x, t + \Delta t \mid x', t) \, p(x', t \mid x_0, t_0) \, dx'$$



*   Answers: given an initial state how does probability evolve?

**Fokker-Planck equation**
$$\frac{\partial p(x, t)}{\partial t} = -\frac{\partial}{\partial x} \left[ A_1(x, t) \, p(x, t) \right] + \frac{\partial^2}{\partial x^2} \left[ A_2(x, t) \, p(x, t) \right]$$

* $A_1(x, t)$: drift coefficient
* $A_2(x, t)$: diffusion coefficient de difusión, commonly $\frac{1}{2}\sigma^2$).

* Answers: given an initial state what is the prob distribution of future states?

**Relation between them**: equivalent, forward and backward formulation

### 1. Financial option
 I pay today a prime(?) to have the right to buy in the future at a set price. I will have the option to buy or not when the time comes. The sellers earns money (my prime) now.


*   Call option: I will have the right to buy the asset in the future at a set price. The seller will have the obligation of selling it to me.
*   Put option: the holder (the option buyer) has the right to sell the asset at a given price. If I decide to exert my right to sell, the other part must buy it.

* American option: can be exercised at any given point in time until the set date.
* European option: only exercised at the set date.
