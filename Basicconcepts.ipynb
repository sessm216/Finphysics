### Lognormal distribution (revising)

What is it= Continuous probability distribution of a random variable whose logarithm is normally distributed. If X is lognormal, means:
$$ Y= lnX \sim N(\mu, \sigma^2) $$


Density function:
$$f(x; \mu, \sigma) = \frac{1}{x \sigma \sqrt{2\pi}} \exp\left(-\frac{(\ln x - \mu)^2}{2\sigma^2}\right)$$

Key facts:
*   takes only positive real values

Sigma meaning: meaning of standard deviation


*   Measures volatility/uncertainty
*   Normal dist: changes the width of the curve. Lognormal dist: changes its form
* $\sigma → 0$: closer to your typical normal curve. Low volatility, very stable
* $\sigma >> 1$: long tail to the right. High risk scenario.  






```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm


x = np.linspace(0.01, 5.0, 500) ## index set positive

# 3 sigma values to compare
mu = 0.0
sigmas = [0.3, 0.6, 1.0]

plt.figure(figsize=(10, 6))

for sigma in sigmas:
    # s: standard deviation of log
    # scale = exp(mu)
    scale = np.exp(mu)
    pdf = lognorm.pdf(x, s=sigma, scale=scale)
    plt.plot(x, pdf, label=f'$\\sigma$ = {sigma}', linewidth=2.5)

plt.title('Comparación $\\sigma$ ($\\mu = 0$)', fontsize=14)
plt.xlabel('Valores de la variable ($X$)', fontsize=12)
plt.ylabel('Densidad de Probabilidad $f(x)$', fontsize=12)
plt.axvline(np.exp(mu), color='gray', linestyle='--', alpha=0.7, label=f'Mediana ($e^\\mu$ = {np.exp(mu)})')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)
plt.ylim(bottom=0)

plt.tight_layout()
plt.show()
```

# Probability distribution in quant finance
Probability density function:
$$ P(a \le X \le b)= ∫_a^bP(x)dx $$


Skewness $S$:
$$
S=
\langle (x-m)^3/\sigma^3 \rangle
$$
About symmetry with respect to the mean value.
Kurtosis $K$:
$$ K = \langle (x-m)^4/\sigma^4 \rangle $$
Characterizes how peaky the peak is. Measurement of how much we deviate from normal distribution

## Stochastic process
Mathematical object usually defined as a family of random variables in a

*   Mathematical object usually defined as a family of random variables in a probability space.
*   Collection of random variables that is indexed by some mathematical set (index set, e.g. integers -> time), meaning that each random variable of the stochastic process is uniquely associated with an element in the set.

*   Each random variable (element) in the collection takes values from the same mathematical space known as the state space.
*   Increment: amount that the stochastic process changes betwen two index values (step), usually two points in time.
*   Sample function/realization: a single outcome of the stochastic process.
